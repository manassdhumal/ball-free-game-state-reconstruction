# Ball-free robustness benchmark

This notebook runs the configured CLEAN, SYNTHETIC-DEGRADED, and NOISE-MITIGATED conditions. Metrica events are used only after inference for evaluation. CLEAN is a TAS reference, not tactical ground truth.

```python
from pathlib import Path
import sys

root = Path.cwd().resolve().parents[1] if Path.cwd().name == '05_tactical_validation' else Path.cwd()
if str(root) not in sys.path:
    sys.path.insert(0, str(root))

from src.validation.robustness_benchmark import load_config, run_benchmark
config_path = root / 'configs' / 'robustness_benchmark.yaml'
config = load_config(config_path)
config
```

```python
result = run_benchmark(config, root=root, save=True)
result['summary_table']
```

```python
import pandas as pd

pd.DataFrame([
    {'condition': name, **summary['tracking_quality_metrics']}
    for name, summary in result['condition_summaries'].items()
])
```

```python
# Optional inline plot when matplotlib is available. A durable SVG/PNG is always saved under results/figures.
try:
    import matplotlib.pyplot as plt
    for condition, scores in result['tactical_scores'].items():
        plt.plot(scores['frame'], scores['tactical_score'], label=condition, linewidth=0.8)
    plt.xlabel('Frame'); plt.ylabel('TAS'); plt.legend(); plt.show()
except ImportError:
    print('matplotlib is unavailable; see the saved SVG comparison figure instead.')
```
